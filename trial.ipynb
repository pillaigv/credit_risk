```python
import pandas as pd
import numpy as np

# Load dataset (adjust path if using Colab or a subfolder like 'data/cs-training.csv')
df = pd.read_csv('Data/cs-training.csv')

# Drop the unneeded index column if Kaggle exported an extra 'Unnamed: 0' column
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# Check the first 5 rows
df.head()
```

```python
# 1. Dataset dimensions
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")

# 2. Column data types & missing value counts
print("=== Data Info & Missing Values ===")
df.info()

# 3. Quantify target class imbalance (SeriousDlqin2yrs)
print("\n=== Target Variable Distribution ===")
default_counts = df['SeriousDlqin2yrs'].value_counts(normalize=True) * 100
print(f"Non-Default (0): {default_counts[0]:.2f}%")
print(f"Default (1):     {default_counts[1]:.2f}%")
```

```python
# Check missing value percentage per feature
missing_pct = (df.isnull().sum() / len(df)) * 100
print("=== Missing Values (%) ===")
print(missing_pct[missing_pct > 0])

# Check basic summary stats for extreme outliers (look at max values for RevolvingUtilization)
df.describe().T[['mean', 'std', 'min', '50%', 'max']]
```
